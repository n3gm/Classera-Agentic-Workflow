```python
# استدعاء المكتبات الخاصة بقراءة الملفات وتقطيعها
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. تحديد مسار الملف (عدلي الاسم حسب الملف اللي حطيتيه في المجلد)
pdf_path = "data/Curriculum_PDFs/test.pdf" 

# 2. قراءة الملف
print("جاري قراءة الكتاب...")
loader = PyPDFLoader(pdf_path)
pages = loader.load()
print(f"✅ تمت قراءة {len(pages)} صفحة بنجاح!")

# 3. تقطيع النص (Chunking)
# نحدد حجم الجزء (1000 حرف) والتداخل بينهم (200 حرف) عشان ما تنقطع الجمل في النص
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=200,
    separators=["\n\n", "\n", " ", ""]
)

chunks = text_splitter.split_documents(pages)
print(f"✅ تم تقسيم الكتاب إلى {len(chunks)} جزء (Chunk) جاهزة للذكاء الاصطناعي.")

# 4. طباعة عينة من أول جزء للتأكد من جودة النص
if chunks:
    print("\n--- عينة من النص المستخرج ---")
    print(chunks[0].page_content)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

print("جاري تحميل نموذج المتجهات (HuggingFace)...")
# نستخدم نموذج مجاني يدعم النصوص متعددة اللغات (بما فيها العربية)
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

print("جاري تحويل النصوص لأرقام وبناء قاعدة البيانات (قد يستغرق بضع ثوانٍ)...")
# إنشاء قاعدة البيانات المتجهة وحفظها في مجلد chroma_db
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="./chroma_db"  # هذا المجلد أضفناه سابقاً في .gitignore لحمايته
)

print("✅ تم تحويل المنهج وتخزينه بنجاح! قاعدة البيانات جاهزة لاستقبال أسئلة الوكلاء.")
```
